```python
# Import pandas
import pandas as pd

# Load the dataset
df = pd.read_csv("airlines_flights_data.csv")

# Display first 5 rows
print("First 5 Rows:")
print(df.head())

# Display dataset information
print("\nDataset Information:")
df.info()

# Display statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Display shape of dataset
print("\nShape of Dataset:")
print(df.shape)

# Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check for duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())

# Display data types of all columns
print("\nData Types:")
print(df.dtypes)
```

```python
# Import libraries
import matplotlib.pyplot as plt

# Histograms for numerical columns
df.hist(figsize=(10, 8))
plt.show()

# Count plot for Airline
df['airline'].value_counts().plot(kind='bar')
plt.title("Airline Count")
plt.xlabel("Airline")
plt.ylabel("Count")
plt.show()

# Count plot for Class
df['class'].value_counts().plot(kind='bar')
plt.title("Travel Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()

# Count plot for Stops
df['stops'].value_counts().plot(kind='bar')
plt.title("Stops")
plt.xlabel("Stops")
plt.ylabel("Count")
plt.show()

# Histogram for Duration
df['duration'].plot(kind='hist', bins=20)
plt.title("Flight Duration Distribution")
plt.xlabel("Duration")
plt.show()

# Histogram for Price
df['price'].plot(kind='hist', bins=20)
plt.title("Ticket Price Distribution")
plt.xlabel("Price")
plt.show()
```

```python
# Task 3: Bivariate Analysis

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("airlines_flights_data.csv")

# Set plot style
sns.set(style="whitegrid")

# 1. Airline vs Price
plt.figure(figsize=(10,5))
sns.boxplot(x='airline', y='price', data=df)
plt.xticks(rotation=45)
plt.title("Airline vs Ticket Price")
plt.show()

# 2. Class vs Price
plt.figure(figsize=(6,5))
sns.boxplot(x='class', y='price', data=df)
plt.title("Travel Class vs Ticket Price")
plt.show()

# 3. Stops vs Price
plt.figure(figsize=(6,5))
sns.boxplot(x='stops', y='price', data=df)
plt.title("Stops vs Ticket Price")
plt.show()

# 4. Duration vs Price
plt.figure(figsize=(8,5))
sns.scatterplot(x='duration', y='price', data=df)
plt.title("Flight Duration vs Ticket Price")
plt.show()

# 5. Days Left vs Price
plt.figure(figsize=(8,5))
sns.scatterplot(x='days_left', y='price', data=df)
plt.title("Days Left vs Ticket Price")
plt.show()
```

```python
# Encode categorical columns
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in df.select_dtypes(include='object').columns:
    df[col] = le.fit_transform(df[col])

# Correlation Heatmap
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.show()

# Feature Engineering
df['price_per_hour'] = df['price'] / df['duration']
df['total_travel_factor'] = df['duration'] * df['stops']
```

```python
# ==========================
# Task 5: Business Insights
# ==========================

print("Top 5 Business Insights")
print("-" * 40)

# 1. Average ticket price by airline
print("\n1. Average Ticket Price by Airline")
print(df.groupby("airline")["price"].mean().sort_values(ascending=False))

# 2. Average ticket price by class
print("\n2. Average Ticket Price by Class")
print(df.groupby("class")["price"].mean())

# 3. Average ticket price by number of stops
print("\n3. Average Ticket Price by Stops")
print(df.groupby("stops")["price"].mean())

# 4. Average ticket price by departure time
print("\n4. Average Ticket Price by Departure Time")
print(df.groupby("departure_time")["price"].mean())

# 5. Correlation with ticket price
print("\n5. Correlation with Price")
print(df.corr(numeric_only=True)["price"].sort_values(ascending=False))

print("\n" + "="*50)
print("Business Recommendations")
print("="*50)

print("1. Encourage early booking by offering discounts to customers who book several days in advance.")

print("2. Optimize pricing based on flight duration and demand to maximize revenue.")

print("3. Improve customer satisfaction by reducing unnecessary stops and offering better services on long-duration flights.")
```
