```python
# Practical 3 - User Defined Functions

# 1. No arguments, No return value
def greet():
    print("Hello!")

greet()

# 2. Arguments, No return value
def add(a, b):
    print("Sum =", a + b)

add(10, 20)

# 3. No arguments, Return value
def message():
    return "Welcome to Python"

print(message())

# 4. Arguments, Return value
def multiply(a, b):
    return a * b

print("Product =", multiply(5, 4))

# 5. Default argument
def student(name="Dharshini"):
    print("Name:", name)

student()
student("Ravi")
# 6. Keyword arguments
def details(name, age):  
    print("Name:", name)
    print("Age:", age)

details(age=18, name="Anu")


# 7. Variable length arguments
def total(*marks):
    print("Total Marks =", sum(marks))

total(80, 90, 85)

# 8. Lambda function
square = lambda x: x * x
print("Square =", square(5))

# 9. Recursive function
def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)

print("Factorial =", fact(5))

# 10. Nested function
def outer():
    def inner():
        print("Inner Function")
    inner()

outer()
```

```python
# Practical 4: Lists and List Comprehensions

# Create a list containing integer elements
numbers = [10, 20, 30, 40, 50]

# Display the original list
print("Original list:", numbers)

# Access and display the first and last elements
print("First element:", numbers[0])
print("Last element:", numbers[-1])

# Add a new element using append()
numbers.append(60)
print("After append:", numbers)

# Insert an element at a specified position
numbers.insert(2, 25)
print("After insert:", numbers)

# Remove an element using remove()
numbers.remove(30)
print("After remove:", numbers)

# Find and display the length of the list
print("Length of the list:", len(numbers))

# Traverse the list using for loop
print("Elements of the list:")
for element in numbers:
    print(element)

# List of squares using list comprehension
squares = [x ** 2 for x in numbers]
print("Squares:", squares)

# List of even numbers using list comprehension
even_numbers = [x for x in numbers if x % 2 == 0]
print("Even numbers:", even_numbers)

# List of odd numbers using list comprehension
odd_numbers = [x for x in numbers if x % 2 != 0]
print("Odd numbers:", odd_numbers)

# Double each element using list comprehension
doubled = [x * 2 for x in numbers]
print("Doubled list:", doubled)

# Create a nested list (matrix)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

print("Matrix:", matrix)

# Flatten the nested list using nested list comprehension
flattened = [x for row in matrix for x in row]
print("Flattened list:", flattened)
```

```python
# practical 12
#Handling Missing Data using Pandas

import pandas as pd
import numpy as np

# Creating DataFrame with missing values
data = {
    "Name": ["Sakshi", "Rahul", "Karthi", "Priya"],
    "Marks": [95, np.nan, 88, np.nan],
    "City": ["Chennai", "Delhi", np.nan, "Mumbai"]
}

df = pd.DataFrame(data)
# Create a DataFrame containing missing values
data = {
    'Name': ['Arun', 'Bala', 'Chitra', 'Divya', 'Elango'],
    'Marks': [85, np.nan, 78, np.nan, 92],
    'Age': [20, 21, np.nan, 22, 21],
    'City': ['Chennai', 'Madurai', np.nan, 'Salem', 'Trichy']
}
df = pd.DataFrame(data)

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Check missing values using isnull()
print("\nMissing values (True/False):")
print(df.isnull())

# Count missing values in each column
print("\nCount of missing values per column:")
print(df.isnull().sum())

# Fill all missing values with a default value using fillna()
df_filled_default = df.fillna("Unknown")
print("\nAfter fillna() with default value 'Unknown':")
print(df_filled_default)

# Fill missing values in Marks column with column mean using fillna()
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
print("\nAfter fillna() - Marks column filled with mean:")
print(df)

# Forward fill using fillna()
df_ffill = df.fillna(method='ffill')
print("\nAfter fillna(method='ffill') - Forward Fill:")
print(df_ffill)

# Backward fill using fillna()
df_bfill = df.fillna(method='bfill')
print("\nAfter fillna(method='bfill') - Backward Fill:")
print(df_bfill)

# Remove rows with missing values using dropna()
df_dropped = df.dropna()
print("\nAfter dropna() - rows with missing values removed:")
print(df_dropped)

# Replace a specified value using replace()
df_replaced = df.replace("Chennai", "Coimbatore")
print("\nAfter replace() - 'Chennai' replaced with 'Coimbatore':")
print(df_replaced)

# Display DataFrame info
print("\nDataFrame Info:")
df.info()
```

```python
# PRACTICAL 18
# Basic Data Cleaning and Preprocessing

import pandas as pd
import numpy as np
import csv

# Loading CSV File
file_path = r"students.csv"


# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display original dataset
print("========== ORIGINAL DATASET ==========")
print(df)

# 3. Display dataset information
print("\n========== ORIGINAL INFORMATION ==========")
df.info()

# 4. Check missing values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# 5. Fill missing Age values with mean
if "Age" in df.columns:
    df["Age"] = df["Age"].fillna(df["Age"].mean())

# 6. Fill missing Marks values with mean
if "Marks" in df.columns:
    df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# 7. Fill missing City values with "Unknown"
if "City" in df.columns:
    df["City"] = df["City"].fillna("Unknown")

# 8. Remove duplicate records
df = df.drop_duplicates()

print("\n========== AFTER REMOVING DUPLICATES ==========")
print(df)

# 9. Remove extra spaces from Name column
if "Name" in df.columns:
    df["Name"] = df["Name"].str.strip()

# 10. Convert Name column to lowercase
if "Name" in df.columns:
    df["Name"] = df["Name"].str.lower()

# 11. Convert City column to lowercase
if "City" in df.columns:
    df["City"] = df["City"].str.lower()

# 12. Convert Age into integer
if "Age" in df.columns:
    df["Age"] = df["Age"].astype(int)

# 13. Rename Marks column as Score
if "Marks" in df.columns:
    df.rename(columns={"Marks": "Score"}, inplace=True)

# 14. Sort dataset by Score
if "Score" in df.columns:
    df = df.sort_values(by="Score", ascending=False)

# 15. Reset the index
df = df.reset_index(drop=True)

# 16. Display cleaned dataset
print("\n========== CLEANED DATASET ==========")
print(df)

# 17. Check missing values after cleaning
print("\n========== MISSING VALUES AFTER CLEANING ==========")
print(df.isnull().sum())

# 18. Display final dataset information
print("\n========== FINAL DATASET INFORMATION ==========")
df.info()
```

```python
# PRACTICAL 19
# Calculate Basic Statistics from a Dataset

import pandas as pd
import numpy as np
import csv

# Loading CSV File
file_path = r"students.csv"


# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display the dataset
print("========== DATASET ==========")
print(df)

# 3. Select Marks column
marks = df["Marks"]

# 4. Remove missing values
marks = marks.dropna()

# 5. Count number of observations
count = marks.count()

# 6. Calculate total marks
total = marks.sum()

# 7. Calculate Mean
mean = marks.mean()

# 8. Calculate Median
median = marks.median()

# 9. Calculate Mode
mode = marks.mode()

# 10. Calculate Standard Deviation
std = marks.std()

# 11. Calculate Variance
variance = marks.var()

# 12. Find Minimum
minimum = marks.min()

# 13. Find Maximum
maximum = marks.max()

# 14. Calculate Range
data_range = maximum - minimum

# 15. Calculate First Quartile
q1 = marks.quantile(0.25)

# 16. Calculate Third Quartile
q3 = marks.quantile(0.75)

# 17. Display results
print("\n========== BASIC STATISTICS ==========")

print("Number of Observations :", count)
print("Total Marks            :", total)
print("Mean                   :", mean)
print("Median                 :", median)
print("Mode                   :", mode.tolist())
print("Standard Deviation     :", std)
print("Variance               :", variance)
print("Minimum                :", minimum)
print("Maximum                :", maximum)
print("Range                  :", data_range)
print("First Quartile (Q1)    :", q1)
print("Third Quartile (Q3)    :", q3)
```
